# JupyDo self-test

Checks that **this environment** works: its kernels (each one is started), sudo, compilers, storage, network
and — when the environment has them — Apptainer, the job queue and GPUs. It adapts to what is here: what
does not apply says **SKIP** with the reason.

| Status | Meaning |
|---|---|
| **PASS** | works |
| **WARN** | works partly, or could not be verified (e.g. no free GPU in the queue right now) |
| **SKIP** | not available in this environment: not an error |
| **FAIL** | something that should work does not: send the output to your JupyDo administrator |

The test is the `jupydo-selftest` command: this notebook just runs it. In an environment without a Python
kernel (e.g. an image with only R), run it in a terminal (*File → New → Terminal*):

    jupydo-selftest

Options: `--no-jobs` (do not submit jobs), `--no-gpu`, `--no-apptainer`, `--no-network`,
`--wait MINUTES` (how long to wait for each job, default 5).
